```python
#Whenever we want to modify any data, 1st copy the data
cities = df["city"].copy() 
cities[0] = "Kolkata" #Modify this into copy not actual data
cities
```

```python
#filtering of data
df[df["aqi"] > 100]
```

```python
#filtering of data
df[(df["aqi"] > 100) & (df["temperature"] > 30)]
```

```python
#chaining
df[df["aqi"] > 100][["city", "aqi"]]
```

```python
df[(df["aqi"] > 100) & (df["temperature"] > 30)][["city", "aqi"]]
```

```python
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] > 30)][["city", "aqi"]]
aqi_data.iloc[0]
```

```python
#Query - copy
df.query("aqi > 100 and temperature > 30")[["city", "aqi"]]
```

```python
# @ reference python vars
aqi_val = 100
df.query("aqi > @aqi_val and temperature > 30")[["city", "aqi"]]
```
