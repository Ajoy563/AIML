```python
#Transforming Data (Frature Engineering)
#Apply
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 50000 else "10%")
df2
```

```python
#Assign()
df2 = df2.assign(new_income = df2["income"] * 1.1)
df2
```

```python
#replace()
df2["country"] = df2["country"].replace("USA", "US")
df2
```

```python
# Melt & Pivot
```

```python
df = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" : [100, 120, 90, 110],
    "profit" : [20, 25, 18, 22], 
    "tax" : [10, 20, 30, 40]
})
df
```

```python
# just like Cartesian Product
melted_df = df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit"],
    var_name = "metrics", 
    value_name = "value"
)
melted_df
```

```python
original = melted_df.pivot(
    index = ["country", "year"],
    columns = "metrics",
    values = "value"
)
original
```

```python
# Merging & Joining
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
```

```python
df_customers
```

```python
df_orders
```

```python
# Inner Join
pd.merge(df_customers, df_orders, on = "customer_id") 
```

```python
# Left Join
pd.merge(df_customers, df_orders, on = "customer_id", how = "left") 
```

```python
# Right Join
pd.merge(df_customers, df_orders, on = "customer_id", how = "right") 
```

```python
# Outer Join
pd.merge(df_customers, df_orders, on = "customer_id", how = "outer") 
```

```python
# Data Concatenation

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Adam", "Bob", "Charlie"]
})

df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name": ["David", "Eva", "Frank"]
})

df1
```

```python
df2
```

```python
pd.concat([df1, df2]) #Row wise
```

```python
pd.concat([df2, df1]) #Row wise
```

```python
pd.concat([df1, df2], ignore_index=True) #Row wise
```

```python
pd.concat([df1, df2], axis=1) #Col wise
```
